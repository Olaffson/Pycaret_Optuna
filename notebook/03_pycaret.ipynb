```python
import pandas as pd
from pycaret.regression import (
    compare_models,
    create_model,
    evaluate_model,
    finalize_model,
    predict_model,
    setup,
    tune_model,
)
```

```python
df = pd.read_csv("../data/gold.csv")
```

```python
reg_setup = setup(data=df, target="prix_median", train_size=0.8, session_id=42)
```

```python
best_model = compare_models()
```

```python
# Évaluation sur le jeu de test (hold-out) mis de côté par setup()
holdout_predictions = predict_model(best_model)
```

```python
# Réentraînement sur l'ensemble des données, une fois l'évaluation faite
final_model = finalize_model(best_model)
```
